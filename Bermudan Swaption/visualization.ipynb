This notebook will be used only in order to visualize the data.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Charger les données générées par le C++
df = pd.read_csv('interpolated_curve.csv')

# Tracer la courbe
plt.figure(figsize=(10, 5))
plt.plot(df['Maturity'], df['Rate'], marker='o', linestyle='-', color='b', label='US Treasury Yield Curve')
plt.title('Courbe des taux US (Interpolée depuis C++)')
plt.xlabel('Maturité (années)')
plt.ylabel('Taux (%)')
plt.grid(True)
plt.legend()
plt.show()
```
